```python
# Creating a Pandas Series
import pandas as pd
marks = pd.Series([85, 90, 78, 92, 88])
print(marks)
```

```python
# Creating a Series with custom index labels
import pandas as pd
marks = pd.Series([85, 90, 78, 92], index=["A", "B", "C", "D"])
print(marks)
```

```python
# Accessing Series elements using index
import pandas as pd
marks = pd.Series([85, 90, 78, 92, 88])
print("First Element:", marks[0])
print("Third Element:", marks[2])
print("Last Element:", marks.iloc[-1])
```

```python
# Filtering Series values
import pandas as pd
marks = pd.Series([85, 45, 90, 32, 78, 95])
passed = marks[marks >= 50]
print("Passing Marks:")
print(passed)
```

```python
# Performing arithmetic operations on a Series
import pandas as pd
marks = pd.Series([10, 20, 30, 40, 50])
print("Addition:", marks + 5)
print("Subtraction:", marks - 5)
print("Multiplication:", marks * 2)
print("Division:", marks / 2)
```

```python
# Calculating statistical values
import pandas as pd
marks = pd.Series([85, 90, 78, 92, 88])
print("Sum:", marks.sum())
print("Mean:", marks.mean())
print("Median:", marks.median())
print("Maximum:", marks.max())
print("Minimum:", marks.min())
```

```python
# Creating a Series from a dictionary
import pandas as pd
student_marks = {"Rahul": 85, "Priya": 90, "Amit": 78, "Sneha": 92}
marks = pd.Series(student_marks)
print(marks)
```

```python
# Checking Series information
import pandas as pd
marks = pd.Series([85, 90, 78, 92, 88])
print("Data Type:", marks.dtype)
print("Number of Elements:", marks.size)
print("Index:", marks.index)
print("Values:", marks.values)
```

```python
# Handling missing values in a Series
import pandas as pd
marks = pd.Series([85, 90, None, 78, None, 92])
print("Original Series:")
print(marks)
print("Missing Values:")
print(marks.isnull())
print("Series Without Missing Values:")
print(marks.dropna())
```

```python
# Sorting Series values
import pandas as pd
marks = pd.Series([85, 45, 90, 32, 78])
print("Ascending Order:")
print(marks.sort_values())
print("Descending Order:")
print(marks.sort_values(ascending=False))
```
